```python
# ============================================================
# STUDENT PERFORMANCE PREDICTOR
# Google Colab Version
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, LogisticRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# 1. DOWNLOAD UCI DATASET
# ============================================================

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00320/student.zip"

# Read ZIP file directly
import zipfile
import requests
import io

response = requests.get(url)

with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    print("Files in dataset:")
    print(z.namelist())

    # Load student-mat.csv
    with z.open("student-mat.csv") as f:
        df = pd.read_csv(f, sep=";")

print("\nDataset loaded successfully!")

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

# ============================================================
# 2. SELECT IMPORTANT FEATURES
# ============================================================

# UCI dataset:
# studytime = weekly study time
# absences  = number of school absences
# G1/G2     = previous grades
# G3        = final grade

# We will use:
# studytime
# absences
# G1
# G2

features = [
    "studytime",
    "absences",
    "G1",
    "G2"
]

target = "G3"

X = df[features]
y = df[target]

print("\nSelected features:")
print(X.head())

# ============================================================
# 3. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")
print(X.isnull().sum())

print("\nTarget missing values:")
print(y.isnull().sum())

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

# Handle missing values
imputer = SimpleImputer(strategy="median")

X_imputed = imputer.fit_transform(X)

# Scale the data
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_imputed)

# ============================================================
# 5. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 6. LINEAR REGRESSION
# ============================================================

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

# Predict final marks
y_pred = linear_model.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("\n================================")
print("LINEAR REGRESSION RESULTS")
print("================================")

print("MAE :", round(mae, 2))
print("MSE :", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 2))

# ============================================================
# 7. ACTUAL VS PREDICTED GRAPH
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    color="blue",
    alpha=0.7
)

plt.xlabel("Actual Final Marks")
plt.ylabel("Predicted Final Marks")
plt.title("Actual vs Predicted Final Marks")

# Reference line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="red",
    linestyle="--"
)

plt.grid()
plt.show()

# ============================================================
# 8. CREATE PASS / FAIL TARGET
# ============================================================

# In the UCI dataset:
# G3 >= 10 means Pass
# G3 < 10 means Fail

df["pass_fail"] = np.where(
    df["G3"] >= 10,
    1,
    0
)

print("\nPass/Fail distribution:")
print(df["pass_fail"].value_counts())

# ============================================================
# 9. CLASSIFICATION DATA
# ============================================================

X_class = df[features]

y_class = df["pass_fail"]

# Handle missing data
X_class_imputed = imputer.fit_transform(X_class)

# Scale
X_class_scaled = scaler.fit_transform(X_class_imputed)

# Train/test split
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class_scaled,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)

# ============================================================
# 10. LOGISTIC REGRESSION
# ============================================================

logistic_model = LogisticRegression(
    random_state=42,
    max_iter=1000
)

logistic_model.fit(
    X_train_c,
    y_train_c
)

# Prediction
y_pred_class = logistic_model.predict(X_test_c)

# ============================================================
# 11. CLASSIFICATION EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test_c,
    y_pred_class
)

precision = precision_score(
    y_test_c,
    y_pred_class,
    zero_division=0
)

recall = recall_score(
    y_test_c,
    y_pred_class,
    zero_division=0
)

print("\n================================")
print("LOGISTIC REGRESSION RESULTS")
print("================================")

print("Accuracy :", round(accuracy, 2))
print("Precision:", round(precision, 2))
print("Recall   :", round(recall, 2))

print("\nClassification Report:")

print(
    classification_report(
        y_test_c,
        y_pred_class,
        target_names=["Fail", "Pass"],
        zero_division=0
    )
)

# ============================================================
# 12. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test_c,
    y_pred_class
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Fail", "Pass"],
    yticklabels=["Fail", "Pass"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()

# ============================================================
# 13. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8, 6))

correlation_data = df[
    ["studytime", "absences", "G1", "G2", "G3"]
]

sns.heatmap(
    correlation_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()

# ============================================================
# 14. FEATURE DISTRIBUTIONS
# ============================================================

df[features].hist(
    figsize=(12, 8),
    color="skyblue",
    edgecolor="black"
)

plt.suptitle("Feature Distributions")

plt.show()

# ============================================================
# 15. PASS/FAIL VISUALIZATION
# ============================================================

plt.figure(figsize=(6, 5))

sns.countplot(
    x=df["pass_fail"],
    hue=df["pass_fail"],
    palette=["red", "green"],
    legend=False
)

plt.xticks(
    [0, 1],
    ["Fail", "Pass"]
)

plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.title("Student Pass/Fail Distribution")

plt.show()

# ============================================================
# 16. USER INPUT / NEW STUDENT PREDICTION
# ============================================================

print("\n================================")
print("NEW STUDENT PREDICTION")
print("================================")

studytime = float(
    input("Enter study time (1-4): ")
)

absences = float(
    input("Enter number of absences: ")
)

G1 = float(
    input("Enter first period grade G1 (0-20): ")
)

G2 = float(
    input("Enter second period grade G2 (0-20): ")
)

# Create new student
new_student = pd.DataFrame({
    "studytime": [studytime],
    "absences": [absences],
    "G1": [G1],
    "G2": [G2]
})

# Apply SAME preprocessing
new_student_imputed = imputer.transform(
    new_student
)

new_student_scaled = scaler.transform(
    new_student_imputed
)

# Predict final marks
predicted_marks = linear_model.predict(
    new_student_scaled
)[0]

# Predict pass/fail
predicted_class = logistic_model.predict(
    new_student_scaled
)[0]

print("\n================================")
print("PREDICTION RESULT")
print("================================")

print(
    "Predicted Final Marks:",
    round(predicted_marks, 2)
)

if predicted_class == 1:
    print("Predicted Result: PASS")
else:
    print("Predicted Result: FAIL")
```
